Data Cleaning

```python
# Loading the dataset csv
import pandas as pd
```

```python
df = pd.read_excel(r"/Users/sucifn/Downloads/Learning Python and Pandas/Customer Call List.xlsx")
```

```python
df
```

```python
df = df.drop_duplicates()
df
```

```python
# Delete columns 
df = df.drop(columns = "Not_Useful_Column")
df
```

```python
# Delete inconsistent format for last_name
df["Last_Name"] = df["Last_Name"].str.strip("123._/")
df
```

```python
# Replace inconsistent format for phone_number
df["Phone_Number"]= df["Phone_Number"].str.replace('[^a-zA-Z0-9]','',regex = True)
df
```

```python
df["Phone_Number"] = df["Phone_Number"].astype(str)
df
```

```python
df["Phone_Number"] = df["Phone_Number"].apply(lambda x: str(x)[0:3] + '-' + str(x)[3:6] + '-' + str(x)[6:10])
df
```

```python
df["Phone_Number"] = df["Phone_Number"].str.replace('nan--', '')
df["Phone_Number"] = df["Phone_Number"].str.replace('Na--', '')
df
```

```python
#df["Phone_Number"].apply(Lambda x: x[0:3] + '-' + x[3:6] + '-' + [6:10])
#df["Phone_Number"] = df["Phone_Number"].apply(Lambda x: str(x))
#df["Phone_Number"].apply(Lambda x: x[0:3] + '-' + x[3:6] + '-' + [6:10])
#df["Phone_Number].str.replace('nan--', '')
#df["Phone_Number].str.replace('Nan--', '')
```

```python
# Split columns 
df[["Street_Address", "State", "Zip_Code"]] = df["Address"].str.split(',', n=2, expand = True)
df
```

```python
# Replace inconsistent format for paying_customer
df["Paying Customer"] = df["Paying Customer"].str.replace('Yes','Y')
df["Paying Customer"] = df["Paying Customer"].str.replace('No','N')
df
```

```python
# Replace inconsistent format for do_not_contact
df["Do_Not_Contact"] = df["Do_Not_Contact"].str.replace('Yes','Y')
df["Do_Not_Contact"] = df["Do_Not_Contact"].str.replace('No','N')
df
```

```python
# Replace null
df = df.fillna('')
df
```

```python
df = df.replace('N/a','')
df
```

```python
# Drop column 
for x in df.index:
    if df.loc[x, "Do_Not_Contact"] == 'Y':
        df.drop(x, inplace = True)
df
```

```python
for x in df.index:
    if df.loc[x, "Phone_Number"] == '':
        df.drop(x, inplace = True)
df
```

```python
df = df.reset_index(drop=True)
df
```
